```python
# AI Chatbot with Database Integration
# Created by Cameron Frost

# -----------------------------------
# Create Company Database
# -----------------------------------

import sqlite3

# Connect to database
connection = sqlite3.connect("company_database.db")

cursor = connection.cursor()


# -----------------------------------
# Create Tables
# -----------------------------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    employee_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    role TEXT,
    location TEXT
)
""")


cursor.execute("""
CREATE TABLE IF NOT EXISTS projects (
    project_id INTEGER PRIMARY KEY,
    project_name TEXT,
    status TEXT,
    manager TEXT
)
""")


connection.commit()

print("Database tables created successfully!")


# -----------------------------------
# Insert Sample Data
# -----------------------------------

# Reset sample data during development/testing
# Prevents duplicate records when rerunning the notebook

cursor.execute("DELETE FROM employees")
cursor.execute("DELETE FROM projects")


# Employee data

employees = [
    (1, "Sarah Johnson", "Engineering", "Software Engineer", "Dallas"),
    (2, "Michael Lee", "IT", "Network Specialist", "Austin"),
    (3, "Emily Davis", "Human Resources", "HR Manager", "Houston"),
    (4, "David Wilson", "Engineering", "Data Engineer", "Dallas"),
    (5, "Jessica Brown", "Security", "Cybersecurity Analyst", "Washington")
]


cursor.executemany("""
INSERT INTO employees
VALUES (?, ?, ?, ?, ?)
""", employees)


# Project data

projects = [
    (1, "Cloud Migration", "Active", "Sarah Johnson"),
    (2, "Network Upgrade", "Completed", "Michael Lee"),
    (3, "Security Enhancement", "Active", "Jessica Brown"),
    (4, "Data Platform", "In Progress", "David Wilson")
]


cursor.executemany("""
INSERT INTO projects
VALUES (?, ?, ?, ?)
""", projects)


# Save database changes

connection.commit()

print("Sample data inserted successfully!")

# -----------------------------------
# Test Database Queries
# -----------------------------------

print("Employees in Database:")
print("----------------------")

cursor.execute("SELECT * FROM employees")

employees_result = cursor.fetchall()

for employee in employees_result:
    print(employee)


print("\nProjects in Database:")
print("--------------------")

cursor.execute("SELECT * FROM projects")

projects_result = cursor.fetchall()

for project in projects_result:
    print(project)

# -----------------------------------
# SQL Query Validation
# -----------------------------------

def validate_query(query):

    query = query.strip().upper()

    if not query.startswith("SELECT"):

        return False

    return True


print("SQL validation ready!")


# -----------------------------------
# Secure Database Query Function
# -----------------------------------

def execute_query(query):

    if not validate_query(query):
        return []

    cursor.execute(query)

    results = cursor.fetchall()

    return results


print("Secure query function ready!")

# -----------------------------------
# Improved SQL Generator
# -----------------------------------

# Prototype natural language parser using keyword matching.
# In a production application, this could be replaced with
# an NLP model or LLM for more advanced query understanding.

def generate_sql(user_input):

    user_input = user_input.lower()


    # Show all employees

    if "employee" in user_input and not any(
        department in user_input
        for department in ["engineering", "it", "human resources", "security"]
    ):
        return "SELECT * FROM employees;"


    # Search employees by department

    departments = [
        "engineering",
        "it",
        "human resources",
        "security"
    ]


    for department in departments:

        if department in user_input:

            return f"""
            SELECT *
            FROM employees
            WHERE department = '{department.title()}';
            """


    # Show all projects

    if "project" in user_input and "active" not in user_input and "completed" not in user_input:
        return "SELECT * FROM projects;"


    # Search projects by status

    if "project" in user_input and "active" in user_input:

        return """
        SELECT *
        FROM projects
        WHERE status = 'Active';
        """


    if "project" in user_input and "completed" in user_input:

        return """
        SELECT *
        FROM projects
        WHERE status = 'Completed';
        """


    return None


print("Improved SQL generator ready!")

# -----------------------------------
# Response Formatting
# -----------------------------------

def format_response(results, request_type):

    if len(results) == 0:
        return "No information found."


    response = ""


    if request_type == "employees":

        response += f"I found {len(results)} employees:\n\n"

        for employee in results:
            response += (
                f"- {employee[1]} is a {employee[3]} "
                f"in the {employee[2]} department "
                f"located in {employee[4]}.\n"
            )


    elif request_type == "projects":

        response += f"I found {len(results)} projects:\n\n"

        for project in results:
            response += (
                f"- {project[1]} is currently {project[2]} "
                f"and is managed by {project[3]}.\n"
            )


    return response


print("Response formatter ready!")

# -----------------------------------
# Improved Chatbot Response Function
# -----------------------------------

def chatbot_response(user_input):

    sql_query = generate_sql(user_input)

    if sql_query is None:
        return "I'm sorry, I don't understand that request yet."


    results = execute_query(sql_query)


    if "employee" in user_input.lower():
        return format_response(results, "employees")


    elif "project" in user_input.lower():
        return format_response(results, "projects")


    else:
        return "Request completed."


print("Improved chatbot system ready!")

# -----------------------------------
# Test Chatbot
# -----------------------------------

print("Chatbot Response:")
print("------------------")

print(chatbot_response("Show me employees"))


print("\nChatbot Response:")
print("------------------")

print(chatbot_response("Show me projects"))

# -----------------------------------
# Test Improved Queries
# -----------------------------------

print("Engineering Employees:")
print("---------------------")

print(chatbot_response("Show me employees in Engineering"))


print("\nActive Projects:")
print("----------------")

print(chatbot_response("Show me active projects"))

# -----------------------------------
# Close Database Connection
# -----------------------------------

connection.close()

print("\nDatabase connection closed.")
```
